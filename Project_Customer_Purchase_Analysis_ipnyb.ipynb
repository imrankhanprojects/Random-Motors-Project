```python
# Project: Customer purchase analysis

## Problem:

## A company wants to understand customer purchases and find total revenue, average order value, and top customer.
```

```python
import pandas as pd

data = {
    "Customer": ["Amit", "Neha", "Amit", "Ravi", "Neha", "Imran"],
    "Product": ["Laptop", "Mobile", "Mouse", "Tablet", "Keyboard", "Monitor"],
    "Amount": [60000, 25000, 1000, 18000, 3000, 12000]
}

df = pd.DataFrame(data)

total_revenue = df["Amount"].sum()
average_order_value = df["Amount"].mean()
customer_revenue = df.groupby("Customer")["Amount"].sum()
top_customer = customer_revenue.idxmax()

print("Total Revenue:", total_revenue)
print("Average Order Value:", average_order_value)
print("Revenue by Customer:")
print(customer_revenue)
print("Top Customer:", top_customer)
```
